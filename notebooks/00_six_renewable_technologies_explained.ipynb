# Six Renewable Technologies Explained

Physical background for the six technologies in our dataset. 
We look at three things:

1. **Mechanism**: what drives the generation.
2. **Expected daily and monthly pattern**: what physics and Chile's climate suggest.
3. **Observed**: what the hourly and monthly seasonal plots in `01_data_exploration.ipynb` (sections 3.1 and 3.2) actually show.

## 0. Background Information of Our Data

Our energy generation time series dataset are data collected from Chile's plants. 

**Chile context**
- The country is in the Southern Hemisphere, so summer is Dec–Feb and winter is Jun–Aug.
- It spans very different climate zones, from the Atacama Desert in the north to the rainy south.
- Our plants follow this geography. Solar plants are in the north (around 25–32°S). Hydro and biomass plants are in central-southern Chile (around 33–41°S). Wind plants are spread between them (around 25–38°S).

**Data**
- The data covers 2024 to April 2026, so May–December are only seen in two years.
- Each series is the sum of all plants of that technology. So some patterns may come from a few plants, which might have bias. 

## 1. Biomass

Organic material such as wood residues and agricultural waste is used as fuel to produce heat, often creating steam that drives a turbine and generator. 
*Main drivers*: Fuel availability, plant capacity, maintenance, and operating schedules.

**Expected.**
- *Daily:* roughly flat/stable, because the plant can generate throughout the day
- *Monthly:* weak natural seasonality. Possible effects from harvest or forestry cycles, fuel supply and planned maintenance.

**Observed.**
- *Daily:* not flat. Generation is about 25 MWh at night but drops to about 16 MWh between 09:00 and 17:00. This looks like an operating choice: the plants generate less while solar floods the grid at midday.
- *Monthly:* no repeating annual cycle (the correlation between years is about 0). The big dips (Nov–Dec 2024, Jan 2025) look like outages or operational events. Only 2 plants are included, so one plant's outage changes the whole series.

## 2. Mini Run-of-River Hydropower (Mini-ROR)

It uses the same fundamental principle as regular run-of-river hydropower, but at a smaller installation scale. *Main driver*: Local river streamflow, hydraulic head, and installed plant capacity.

**Expected.**
- *Daily:* fairly smooth, with no dependence on daylight.
- *Monthly:* follows the local hydrology. Rain-fed rivers peak in winter (Jun–Aug), and snowmelt-fed rivers peak in spring and early summer (Oct–Dec). Small streams may react more strongly to local weather.

**Observed.**
- *Daily:* almost flat (25–32 MWh), with a small step down at about 20:00.
- *Monthly:* a **clear and repeating annual cycle** (correlation between years about 0.93). Output is lowest in Mar–Apr, at the end of the dry summer, and rises to a peak in Oct–Dec, which fits spring snowmelt.

## 3. Reservoir Hydropower

Water is stored behind a dam in a reservoir and released through turbines when electricity generation is needed. *Main drivers*: Reservoir water availability, inflows, storage constraints, and electricity-system operation.

**Expected.**
- *Daily:* peaks shaped by dispatch decisions, with generation high when demand is high and solar is low.
- *Monthly:* follows inflow and storage, but water can be shifted in time, so the link to the season is weaker.

**Observed.**
- *Daily:* the strongest dispatch signature in the dataset. Output is about 1,600 MWh in the evening (20:00–22:00) and about 600 MWh at midday (12:00–15:00). Reservoirs hold back water while solar covers the midday hours, then fill the evening demand peak after sunset.
- *Monthly:* some shape, with a dip around Mar–Apr, but it repeats only weakly (correlation about 0.42). A **downward trend** dominates instead: average output falls from about 1,400 MWh in 2024 to about 700 MWh in 2026. This could be declining reservoir storage (drought) or changes in operation.

## 4. Run-of-River Hydropower (ROR)

Water from a river is diverted through turbines, producing electricity using natural streamflow. *Main driver*: River discharge and hydraulic head.

**Expected.**
- *Daily:* relatively smooth. Rivers fed by snow or glaciers can show small daily melt cycles.
- *Monthly:* strong annual seasonality. Central-southern Chilean rivers are rain-fed (winter peak), snowmelt-fed (spring/early-summer peak) or mixed.

**Observed.**
- *Daily:* smooth, but with a mild midday dip (about 500 MWh at night and about 405 MWh at 15:00). This is consistent with some plants having small ponds that shift water toward the evening, or with midday curtailment.
- *Monthly:* a **clear annual cycle that mostly repeats**. Output is lowest in Mar–Apr and highest in Oct–Jan, which matches spring snowmelt. 2025 is lower overall than 2024.

## 5. Solar Photovoltaic (Solar PV)

Solar panels use photovoltaic cells to convert sunlight directly into electricity. *Main driver*: Solar irradiance, influenced by the sun's position, clouds, and weather.

**Expected.**
- *Daily:* a strong, predictable 24-hour cycle. Output is zero at night, rises after sunrise, peaks around midday and falls to zero at sunset.
- *Monthly:* higher in the austral summer (Dec–Feb) and lower in winter (Jun–Aug). Our plants are in the Atacama region, which has very few clouds, so we expect a smooth and stable seasonal curve.

**Observed.**
- *Daily:* exactly as expected. Output is zero from 21:00 to 06:00 and peaks at about 340 MWh between 12:00 and 13:00.
- *Monthly:* a **clear and very repeatable annual cycle** (correlation between years about 0.93). Output peaks in Dec–Feb and is lowest around May–Aug. This is the most predictable technology in the dataset, at both the daily and the annual scale.

## 6. Wind

Wind moves turbine blades, which drive a generator to produce electricity. *Main driver*: Wind speed, with additional influences from turbulence, wind direction, and turbine operating limits.

**Expected.**
- *Daily:* no sunlight is needed, so wind can produce day and night. Any daily cycle depends on the site: daytime heating, sea breezes on the coast, or mountain–valley winds.
- *Monthly:* depends on regional circulation. Chile spans many climate zones, so there is no universal peak month.

**Observed.**
- *Daily:* a clear average cycle. The low is about 90 MWh at about 09:00 and the high is about 228 MWh at about 20:00. This fits afternoon/evening winds driven by heating along the coast and in the north. The mean hides large day-to-day scatter.
- *Monthly:* **no stable annual cycle** (correlation between years about 0.23). 2024 was higher in May–Sep, while 2025 was almost flat. Weather variability matters more than the season.
